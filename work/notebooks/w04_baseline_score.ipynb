## 4. Weak picks + leakage check

I challenge the baseline by identifying weak recommendations, such as <b> old pages with very little traffic, pages with short-term declines, or pages with insufficient history. </b> I also perform a leakage audit to ensure that future clicks, impressions, rankings, post-refresh results, future product metrics, and other information unavailable at prediction time are not used as features. This ensures that the baseline is a fair and realistic benchmark for the eventual ML model.

```python
forbidden_terms = [
    "future",
    "next",
    "post_refresh",
    "after_refresh",
    "refresh_result",
    "final_outcome",
    "future_window"
]

feature_columns = [
    "clicks",
    "impressions",
    "ctr",
    "avg_position",
    "content_age_days",
    "click_change_pct",
    "impression_change_pct",
    "ctr_change_pct",
    "position_change",
    "history_periods"
]

suspicious = [
    col for col in feature_columns
    if any(term in col.lower() for term in forbidden_terms)
]

print("Suspicious feature columns:", suspicious)

assert len(suspicious) == 0, (
    "Potential leakage detected in feature columns."
)
```
