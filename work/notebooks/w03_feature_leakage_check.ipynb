## 3. The leakage hunt

The main leakage risk is allowing future performance information to enter the feature vector. The proxy label is based on future performance, such as the change in clicks during the following 30-day period. Therefore, future clicks, impressions, CTR, position, engagement, post-refresh performance, and final outcomes are excluded from the features. Rolling features are calculated using a one-period shift so that the current or future observation cannot enter the historical average. I also check feature names and the dataframe for future, post-intervention, outcome, or manually assigned decision fields. ## 4. What I excluded and why

I exclude future clicks, impressions, CTR, average position, and engagement because they contain information that would not be available when making the prediction. Post-refresh performance and refresh outcomes are excluded because they occur after the recommendation and can directly leak the outcome. Manually assigned opportunity scores and actions are excluded because they may already contain human judgment that the model is supposed to learn. Raw URLs and page IDs are excluded as direct model features because they are identifiers rather than generalizable behavioral signals and can cause the model to memorize individual pages. Any field collected after the prediction timestamp is also excluded to maintain a strict prediction-time information boundary.
