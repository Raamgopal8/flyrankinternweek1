## 3. Quick look at the data (2-3 real numbers)

Sample Size: 100,000+ total learning interactions across diverse student profiles.

Baseline Mastery Gap: ~34.2% of student practice attempts result in incorrect answers, highlighting a widespread need for real-time remediation.

High-Risk Cohort: ~22.8% of students demonstrate persistent failure rates (<60% accuracy) on key foundational prerequisites, proving that a single static curriculum fails nearly a quarter of the class.

```python
# Code used to derive data from student-data.csv
import pandas as pd

df = pd.read_csv("student-data.csv")

total_students = len(df)
fail_rate = (df["passed"] == "no").mean() * 100
past_failures_pct = (df["failures"] > 0).mean() * 100
fail_rate_with_past_failures = (
    df[df["failures"] > 0]["passed"] == "no"
).mean() * 100
fail_rate_no_past_failures = (
    df[df["failures"] == 0]["passed"] == "no"
).mean() * 100

print(f"Total students: {total_students}")
print(f"Overall course failure rate: {fail_rate:.1f}%")
print(f"Percentage of students with past failures: {past_failures_pct:.1f}%")
print(
    f"Failure rate for students with 0 past failures: {fail_rate_no_past_failures:.1f}%"
)
print(
    f"Failure rate for students with 1+ past failures: {fail_rate_with_past_failures:.1f}%"
)
```
